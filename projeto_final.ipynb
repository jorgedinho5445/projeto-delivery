```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("delivery_pedidos.csv")

print(df.shape)

df.head()
import pandas as pd
from scipy import stats

df = pd.read_csv('delivery_pedidos_limpo.csv')
print("=== TAREFA 11.1: DICIONÁRIO DE DADOS ===")
print(f"Tamanho final: {df.shape[0]} linhas x {df.shape[1]} colunas\n")
for col in df.columns:
    faltantes = (df[col].isnull().sum() / len(df)) * 100
    print(f"| `{col}` | {df[col].dtype} | — | {faltantes:.1f}% | — |")

print("\n=== TAREFA 9.7: TESTES DE HIPÓTESE (α = 5%) ===")
sem_chuva = df[df['chuva'] == False]['tempo_entrega_min'].dropna()
com_chuva = df[df['chuva'] == True]['tempo_entrega_min'].dropna()
t_h1, p_h1 = stats.ttest_ind(com_chuva, sem_chuva)
print(f"H1 (Chuva x Tempo): dif = +{com_chuva.mean() - sem_chuva.mean():.1f} min | p-valor = {p_h1:.4e}")

if 'usou_cupom' in df.columns:
    sem_cupom = df[df['usou_cupom'] == False]['valor_pedido'].dropna()
    com_cupom = df[df['usou_cupom'] == True]['valor_pedido'].dropna()
    t_h2, p_h2 = stats.ttest_ind(com_cupom, sem_cupom)
    print(f"H2 (Cupom x Ticket): dif = R$ {com_cupom.mean() - sem_cupom.mean():.2f} | p-valor = {p_h2:.4e}")

df_h3 = df[['tempo_entrega_min', 'avaliacao']].dropna()
r_h3, p_h3 = stats.spearmanr(df_h3['tempo_entrega_min'], df_h3['avaliacao'])
print(f"H3 (Tempo x Avaliação): corr = {r_h3:.3f} | p-valor = {p_h3:.4e}")

if 'tipo_cozinha' in df.columns:
    grupos_cozinha = [g['valor_pedido'].dropna() for _, g in df.groupby('tipo_cozinha')]
    f_h4, p_h4 = stats.f_oneway(*grupos_cozinha)
    print(f"H4 (Cozinha x Ticket): F = {f_h4:.2f} | p-valor = {p_h4:.4e}")

if 'forma_pagamento' in df.columns:
    tabela_h5 = pd.crosstab(df['cidade'], df['forma_pagamento'])
    chi2_h5, p_h5, _, _ = stats.chi2_contingency(tabela_h5)
    print(f"H5 (Cidade x Pagamento): chi2 = {chi2_h5:.2f} | p-valor = {p_h5:.4e}")

df_h6 = df[['distancia_km', 'tempo_entrega_min']].dropna()
r_h6, p_h6 = stats.pearsonr(df_h6['distancia_km'], df_h6['tempo_entrega_min'])
print(f"H6 (Distância x Tempo): corr = {r_h6:.3f} | p-valor = {p_h6:.4e}")

```
